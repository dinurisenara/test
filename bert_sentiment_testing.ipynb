```python
from transformers import BertForSequenceClassification, BertTokenizer
import torch
```

```python
# Define model path
model_path = "/content/drive/MyDrive/bert_sentiment_model"

# Load the fine-tuned BERT model and tokenizer
model = BertForSequenceClassification.from_pretrained(model_path)
tokenizer = BertTokenizer.from_pretrained(model_path)

# Move model to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print("Model loaded successfully!")
```

Define prediction Function

```python
def predict_sentiment(text):
    # Tokenize input text
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=128).to(device)

    # Get model predictions
    with torch.no_grad():
        outputs = model(**inputs)

    # Convert logits to predicted label
    logits = outputs.logits
    predicted_class = torch.argmax(logits, dim=1).item()

    # Convert label to sentiment
    label_mapping = {0: "Negative", 1: "Positive"}
    return label_mapping[predicted_class]
```

Test with real sentences

```python
test_sentences = [
    "I love this product! It's amazing.",
    "I am feeling really sad today.",
    "This is the worst service I have ever experienced.",
    "The movie was absolutely fantastic!",
    "I'm not sure how I feel about this."
]

# Run predictions
for sentence in test_sentences:
    sentiment = predict_sentiment(sentence)
    print(f"Text: {sentence}\nPredicted Sentiment: {sentiment}\n")
```
